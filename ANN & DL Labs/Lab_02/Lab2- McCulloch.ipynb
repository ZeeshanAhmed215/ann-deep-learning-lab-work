# AND Gate

```python
import tensorflow as tf

# Define the AND gate function
def and_gate(input1, input2):
    # Define the inputs as TensorFlow tensors
    inputs = tf.constant([[input1, input2]], dtype=tf.float32)
    
    # Define the weights without bias
    weights = tf.constant([[1.0, 1.0]], dtype=tf.float32)
    
    # Calculate the weighted sum
    weighted_sum = tf.matmul(inputs, tf.transpose(weights))
    
    # Apply the step function with threshold 0
    output = tf.where(weighted_sum >= 2, 1.0, 0.0)
    
    return output.numpy()[0][0]

# Testing the AND gate
print("AND gate output:")
print("0 AND 0 =", and_gate(0, 0))
print("0 AND 1 =", and_gate(0, 1))
print("1 AND 0 =", and_gate(1, 0))
print("1 AND 1 =", and_gate(1, 1))
```

# OR Gate

```python
import tensorflow as tf

# Define the OR gate function
def or_gate(input1, input2):
    # Define the inputs as TensorFlow tensors
    inputs = tf.constant([[input1, input2]], dtype=tf.float32)
    
    # Define the weights without bias
    weights = tf.constant([[1.0, 1.0]], dtype=tf.float32)
    
    # Calculate the weighted sum
    weighted_sum = tf.matmul(inputs, tf.transpose(weights))
    
    # Apply the step function with threshold 0
    output = tf.where(weighted_sum >= 1, 1.0, 0.0)
    
    return output.numpy()[0][0]

# Testing the OR gate
print("OR gate output:")
print("0 OR 0 =", or_gate(0, 0))
print("0 OR 1 =", or_gate(0, 1))
print("1 OR 0 =", or_gate(1, 0))
print("1 OR 1 =", or_gate(1, 1))
```

# ANDNOT

```python
import tensorflow as tf

# Define the ANDNOT gate function
def andnot_gate(input1, input2):
    # Define the inputs as TensorFlow tensors
    inputs = tf.constant([[input1, input2]], dtype=tf.float32)
    
    # Define the weights without bias
    weights = tf.constant([[1.0, -1.0]], dtype=tf.float32)
    
    # Calculate the weighted sum
    weighted_sum = tf.matmul(inputs, tf.transpose(weights))
    
    # Apply the step function with threshold 0
    output = tf.where(weighted_sum >= 1, 1.0, 0.0)
    
    return output.numpy()[0][0]

# Testing the ANDNOT gate
print("ANDNOT gate output:")
print("0 ANDNOT 0 =", andnot_gate(0, 0))
print("0 ANDNOT 1 =", andnot_gate(0, 1))
print("1 ANDNOT 0 =", andnot_gate(1, 0))
print("1 ANDNOT 1 =", andnot_gate(1, 1))
```

# NOR Gate

```python
import tensorflow as tf

# Define the NOR gate function
def nor_gate(input1, input2):
    # Define the inputs as TensorFlow tensors
    inputs = tf.constant([[input1, input2]], dtype=tf.float32)
    
    # Define the weights without bias
    weights = tf.constant([[-1.0, -1.0]], dtype=tf.float32)
    
    # Calculate the weighted sum
    weighted_sum = tf.matmul(inputs, tf.transpose(weights))
    
    # Apply the step function with threshold 0
    output = tf.where(weighted_sum >= 0, 0.0, 1.0)
    
    return output.numpy()[0][0]

# Testing the NOR gate
print("NOR gate output:")
print("0 NOR 0 =", nor_gate(0, 0))
print("0 NOR 1 =", nor_gate(0, 1))
print("1 NOR 0 =", nor_gate(1, 0))
print("1 NOR 1 =", nor_gate(1, 1))
```

# NAND Gate

```python
import tensorflow as tf

# Define the NAND gate function
def nand_gate(input1, input2):
    # Define the inputs as TensorFlow tensors
    inputs = tf.constant([[input1, input2]], dtype=tf.float32)
    
    # Define the weights without bias
    weights = tf.constant([[-1.0, -1.0]], dtype=tf.float32)
    
    # Calculate the weighted sum
    weighted_sum = tf.matmul(inputs, tf.transpose(weights))
    
    # Apply the step function with threshold -1
    output = tf.where(weighted_sum >= -1, 1.0, 0.0)
    
    return output.numpy()[0][0]

# Testing the NAND gate
print("NAND gate output:")
print("0 NAND 0 =", nand_gate(0, 0))
print("0 NAND 1 =", nand_gate(0, 1))
print("1 NAND 0 =", nand_gate(1, 0))
print("1 NAND 1 =", nand_gate(1, 1))
```

# XOR Gate

```python
import tensorflow as tf

# Define the XOR gate function
def xor_gate(input1, input2):
    # Define the inputs as TensorFlow tensors
    inputs = tf.constant([[input1, input2]], dtype=tf.float32)
    
    # Define the weights for the first layer (input layer to hidden layer)
    weights_input_hidden = tf.constant([[1.0, -1.0], [-1.0, 1.0]], dtype=tf.float32)
    
    # Calculate the weighted sum for the first layer
    weighted_sum_input_hidden = tf.matmul(inputs, weights_input_hidden)
    
    # Apply the step function for the first layer
    output_hidden = tf.where(weighted_sum_input_hidden >= 1, 1.0, 0.0)
    
    # Define the weights for the second layer (hidden layer to output layer)
    weights_hidden_output = tf.constant([[1.0], [1.0]], dtype=tf.float32)
    
    # Calculate the weighted sum for the second layer
    weighted_sum_hidden_output = tf.matmul(output_hidden, weights_hidden_output)
    
    # Apply the step function for the second layer
    output = tf.where(weighted_sum_hidden_output >= 1, 1.0, 0.0)
    
    return output.numpy()[0][0]

# Testing the XOR gate
print("XOR gate output:")
print("0 XOR 0 =", xor_gate(0, 0))
print("0 XOR 1 =", xor_gate(0, 1))
print("1 XOR 0 =", xor_gate(1, 0))
print("1 XOR 1 =", xor_gate(1, 1))
```
