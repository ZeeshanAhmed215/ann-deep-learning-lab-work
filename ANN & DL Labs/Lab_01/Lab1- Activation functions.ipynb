# Linear activation function 

The Linear activation function is the simplest form of activation function where the output is directly proportional to the input. It can be represented as:

𝑓(𝑥)=𝑥

The Linear activation function is useful for regression problems, where the goal is to predict a continuous output value. It is also commonly used as the activation function for hidden layers in a neural network, as it allows for simple and efficient computation.

However, the Linear activation function may not be appropriate for all problems, as it may not be able to effectively model the complex relationships between input and output. In such cases, other activation functions such as the ReLU or the sigmoid function may provide better performance.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the Linear activation function
def linear(x):
    return x

# Generate input data
x = np.linspace(-10, 10, 100)

# Apply the Linear activation function
y = linear(x)

# Plot the Linear activation function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='Linear Activation Function')
plt.title('Linear Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid(True)
plt.show()
```

# ReLU (Rectified Linear Unit) activation function

The ReLU (Rectified Linear Unit) activation function is a popular activation function used in neural networks. It is defined as:

𝑓(𝑥)=max(0,𝑥)

The ReLU activation function has several advantages over other activation functions, including:

It allows for faster and more efficient training of neural networks.
It can handle the vanishing gradient problem, which is common in deep neural networks.
It is relatively easy to compute and optimize.
It has been shown to achieve good performance on a variety of tasks, such as image classification and natural language processing.
import numpy as np

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the ReLU activation function
def relu(x):
    return np.maximum(0, x)

# Generate input data
x = np.linspace(-10, 10, 400)

# Apply the ReLU activation function
y = relu(x)

# Plot the ReLU activation function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='ReLU Activation Function')
plt.title('ReLU Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid(True)
plt.show()
```

# Leaky ReLU activation function

Leaky ReLU is an extension of the ReLU activation function. It is similar to ReLU, but instead of returning zero for negative inputs, it returns a small negative value. This helps to avoid the "dying ReLU" problem, where some neurons can become permanently inactive during training.

The Leaky ReLU activation function is defined as:

𝑓(𝑥)=𝑎𝑥,𝑥<0

```python
import numpy as np
import matplotlib.pyplot as plt
# Define the Leaky ReLU activation function
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)
# Generate input data
x = np.linspace(-10, 10, 400)
# Apply the Leaky ReLU activation function
y = leaky_relu(x)
# Plot the Leaky ReLU activation function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='Leaky ReLU Activation Function')
plt.title('Leaky ReLU Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid(True)
plt.show()
```

# Sigmoid or logistic activation function

The Sigmoid activation function is commonly used in neural networks, especially for binary classification tasks. The Sigmoid function maps any real-valued number into the range (0, 1).

The Sigmoid function can be defined as:

𝑓(𝑥)=  1/(1+e^(-x))
​


```python
import numpy as np
import matplotlib.pyplot as plt

# Define the Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Generate input data
x = np.linspace(-10, 10, 400)

# Apply the Sigmoid activation function
y = sigmoid(x)

# Plot the Sigmoid activation function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='Sigmoid Activation Function', color='b')
plt.title('Sigmoid Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid(True)
plt.show()
```

# Bipolar sigmoid activation function

The Bipolar Sigmoid activation function is similar to the standard Sigmoid function but scales the output to the range (-1, 1) instead of (0, 1). It can be defined as:

f(x)=  (1-e^(-x))/(1+e^(-x))

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the Bipolar Sigmoid activation function
def bipolar_sigmoid(x):
    return (1 - np.exp(-x)) / (1 + np.exp(-x))

# Generate input data
x = np.linspace(-10, 10, 400)

# Apply the Bipolar Sigmoid activation function
y = bipolar_sigmoid(x)

# Plot the Bipolar Sigmoid activation function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='Bipolar Sigmoid Activation Function', color='g')
plt.title('Bipolar Sigmoid Activation Function')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid(True)
plt.show()
```
