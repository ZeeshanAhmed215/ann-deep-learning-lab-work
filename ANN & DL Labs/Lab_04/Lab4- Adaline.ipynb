# AND Gate Implementation using Adaline

```python
import numpy as np

def adaline_train(X, y, learning_rate=0.1, epochs=100):
    n_samples, n_features = X.shape
    weights = np.zeros(n_features)
    bias = 0
    
    for _ in range(epochs):
        # Compute the net input (weighted sum)
        net_input = np.dot(X, weights) + bias
        
        # Compute the predicted output (activation function)
        y_pred = np.where(net_input >= 0, 1, 0)
        
        # Compute the errors
        errors = y - y_pred
        
        # Update weights and bias
        weights += learning_rate * np.dot(X.T, errors)
        bias += learning_rate * np.sum(errors)
    
    return weights, bias

def adaline_predict(X, weights, bias):
    net_input = np.dot(X, weights) + bias
    return np.where(net_input >= 0, 1, 0)

# Define the AND gate inputs and outputs
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])

# Train Adaline on the AND gate data
weights, bias = adaline_train(X, y, learning_rate=0.1, epochs=100)

# Test the trained Adaline model on the AND gate inputs
print("AND gate output:")
for i in range(len(X)):
    prediction = adaline_predict(X[i], weights, bias)
    print(f"{X[i][0]} AND {X[i][1]} =", prediction)
```
