# XOR Gate Implementation using Medaline

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def medaline_train(X, y, hidden_units=2, learning_rate=0.1, epochs=1000):
    n_samples, n_features = X.shape
    
    # Initialize weights and biases for the hidden layer
    hidden_weights = np.random.uniform(size=(n_features, hidden_units))
    hidden_bias = np.zeros(hidden_units)
    
    # Initialize weights and bias for the output layer
    output_weights = np.random.uniform(size=(hidden_units, 1))
    output_bias = 0
    
    for _ in range(epochs):
        # Forward pass
        hidden_input = np.dot(X, hidden_weights) + hidden_bias
        hidden_output = sigmoid(hidden_input)
        
        output_input = np.dot(hidden_output, output_weights) + output_bias
        output_output = sigmoid(output_input)
        
        # Backpropagation
        output_error = y - output_output
        output_delta = output_error * sigmoid_derivative(output_output)
        
        hidden_error = np.dot(output_delta, output_weights.T)
        hidden_delta = hidden_error * sigmoid_derivative(hidden_output)
        
        # Update weights and biases
        output_weights += learning_rate * np.dot(hidden_output.T, output_delta)
        output_bias += learning_rate * np.sum(output_delta, axis=0)
        
        hidden_weights += learning_rate * np.dot(X.T, hidden_delta)
        hidden_bias += learning_rate * np.sum(hidden_delta, axis=0)
    
    return hidden_weights, hidden_bias, output_weights, output_bias

def medaline_predict(X, hidden_weights, hidden_bias, output_weights, output_bias):
    hidden_output = sigmoid(np.dot(X, hidden_weights) + hidden_bias)
    output_output = sigmoid(np.dot(hidden_output, output_weights) + output_bias)
    return np.round(output_output)

# Define the XOR gate inputs and outputs
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

# Train Medaline on the XOR gate data
hidden_weights, hidden_bias, output_weights, output_bias = medaline_train(X, y, hidden_units=2, learning_rate=0.1, epochs=10000)

# Test the trained Medaline model on the XOR gate inputs
print("XOR gate output:")
for i in range(len(X)):
    prediction = medaline_predict(X[i], hidden_weights, hidden_bias, output_weights, output_bias)
    print(f"{X[i][0]} XOR {X[i][1]} =", prediction[0])
```
